## Initialising and invoking a model

```python
from langchain.chat_models import init_chat_model

gptLunaModel = init_chat_model(model="gpt-6-luna")
claudeHaikuModel = init_chat_model(model="claude-haiku-5-5")
```

```python
gptLunaResponse = gptLunaModel.invoke("What's the capital of the Moon?")
claudeHaikuResponse = claudeHaikuModel.invoke("What's the capital of the Moon?")

gptLunaResponse
```

```python
print("gptLunaResponse:" + str(gptLunaResponse.content))
print("claudeHaikuResponse:" + str(claudeHaikuResponse.content))
```

```python
from pprint import pprint

# pprint(gptLunaResponse.response_metadata)
pprint(claudeHaikuResponse.response_metadata)
```

## Customising your Model

```python
gptLunaModel = init_chat_model(
    model="gpt-6-luna",
    # Kwargs passed to the model:
    temperature=1.0
)

gptLunaResponse = gptLunaModel.invoke("What's the capital of the Moon?")
print(gptLunaResponse.content)
```

## Initialising and invoking an agent

```python
from langchain.agents import create_agent

agent = create_agent(model=gptLunaModel)
```

```python
agent = create_agent(model="claude-sonnet-4-6")
```

```python
agent = create_agent("gpt-6-luna")
```

```python
from langchain.messages import HumanMessage

gptLunaResponse = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?")]}
)
```

```python
from pprint import pprint

pprint(gptLunaResponse)
```

```python
print(gptLunaResponse['messages'][-1].content)
```

```python
from langchain.messages import AIMessage

gptLunaResponse = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)

print(gptLunaResponse['messages'][-1].content)
```

## Streaming Output

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
